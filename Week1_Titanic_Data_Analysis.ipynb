```python
# Week 1: Data Acquisition, Cleaning, and Exploratory Data Analysis

# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

# Visualization settings
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")

# ============================================================
# 2. LOAD THE DATASET
# ============================================================

# Load Titanic dataset
df = pd.read_csv("train.csv")

print("Dataset loaded successfully.")

# ============================================================
# 2. LOAD THE DATASET
# ============================================================

# Load Titanic dataset
df = pd.read_csv("train.csv")

print("Dataset loaded successfully.")

print("Last 5 rows of the dataset:")
display(df.tail())

# ============================================================
# 4. DATASET DIMENSIONS
# ============================================================

rows, columns = df.shape

print("Number of rows:", rows)
print("Number of columns:", columns)
print("Dataset shape:", df.shape)

# ============================================================
# 5. COLUMN NAMES
# ============================================================

print("Columns in the dataset:")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

# ============================================================
# 6. DATASET INFORMATION
# ============================================================

print("Dataset Information:")
df.info()

# ============================================================
# 7. STATISTICAL SUMMARY
# ============================================================

print("Statistical summary of numerical columns:")
display(df.describe())

# ============================================================
# 8. CATEGORICAL DATA SUMMARY
# ============================================================

print("Summary of categorical columns:")
display(df.describe(include="object"))

# ============================================================
# 9. DATA TYPES
# ============================================================

print("Data types of all columns:")
display(df.dtypes)

# ============================================================
# 10. MISSING VALUE ANALYSIS
# ============================================================

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage
})

missing_table = missing_table.sort_values(
    by="Missing Values",
    ascending=False
)

print("Missing value analysis:")
display(missing_table)

# ============================================================
# 11. VISUALIZATION - MISSING VALUES
# ============================================================

missing_plot = missing_count[missing_count > 0].sort_values(
    ascending=False
)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=missing_plot.index,
    y=missing_plot.values
)

plt.title("Missing Values by Column")
plt.xlabel("Column")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.tight_layout()

# Save figure
plt.savefig("missing_values.png", dpi=300, bbox_inches="tight")

plt.show()

# ============================================================
# 12. DUPLICATE RECORD ANALYSIS
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count == 0:
    print("No duplicate records were found.")
else:
    print("Duplicate records were found and will be removed.")

# ============================================================
# 13. INFINITE VALUE ANALYSIS
# ============================================================

numeric_columns = df.select_dtypes(include=np.number).columns

infinite_values = np.isinf(df[numeric_columns]).sum()

print("Infinite values in numerical columns:")
display(infinite_values)

if infinite_values.sum() == 0:
    print("No infinite values were found.")
else:
    print("Infinite values were found.")

# ============================================================
# 14. CREATE CLEANING COPY
# ============================================================

cleaned_df = df.copy()

print("A copy of the original dataset has been created for cleaning.")

# ============================================================
# 15. HANDLE INFINITE VALUES
# ============================================================

cleaned_df[numeric_columns] = cleaned_df[numeric_columns].replace(
    [np.inf, -np.inf],
    np.nan
)

print("Infinite values have been replaced with NaN where applicable.")

# ============================================================
# 16. HANDLE MISSING AGE VALUES
# ============================================================

age_median = cleaned_df["Age"].median()

cleaned_df["Age"] = cleaned_df["Age"].fillna(age_median)

print("Missing Age values filled using median.")
print("Median Age used:", age_median)

# ============================================================
# 17. HANDLE MISSING EMBARKED VALUES
# ============================================================

embarked_mode = cleaned_df["Embarked"].mode()[0]

cleaned_df["Embarked"] = cleaned_df["Embarked"].fillna(embarked_mode)

print("Missing Embarked values filled using mode.")
print("Mode used:", embarked_mode)

# ============================================================
# 18. HANDLE CABIN COLUMN
# ============================================================

cabin_missing_percentage = (
    cleaned_df["Cabin"].isnull().sum() / len(cleaned_df)
) * 100

print(f"Missing Cabin percentage: {cabin_missing_percentage:.2f}%")

# Remove Cabin because of its high proportion of missing values
cleaned_df = cleaned_df.drop(columns=["Cabin"])

print("Cabin column removed because of extensive missing values.")

# ============================================================
# 19. REMOVE DUPLICATE RECORDS
# ============================================================

before_duplicates = len(cleaned_df)

cleaned_df = cleaned_df.drop_duplicates()

after_duplicates = len(cleaned_df)

print("Rows before duplicate removal:", before_duplicates)
print("Rows after duplicate removal:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)

# ============================================================
# 20. CHECK DATA TYPES AFTER CLEANING
# ============================================================

print("Data types after cleaning:")
display(cleaned_df.dtypes)

# ============================================================
# 21. VERIFY MISSING VALUES AFTER CLEANING
# ============================================================

remaining_missing = cleaned_df.isnull().sum()

print("Remaining missing values:")
display(remaining_missing)

if remaining_missing.sum() == 0:
    print("Excellent! No missing values remain.")
else:
    print("Some missing values still remain.")

# ============================================================
# 22. VERIFY INFINITE VALUES AFTER CLEANING
# ============================================================

numeric_columns_cleaned = cleaned_df.select_dtypes(
    include=np.number
).columns

remaining_infinite = np.isinf(
    cleaned_df[numeric_columns_cleaned]
).sum()

print("Remaining infinite values:")
display(remaining_infinite)

if remaining_infinite.sum() == 0:
    print("No infinite values remain.")

# ============================================================
# 22. VERIFY INFINITE VALUES AFTER CLEANING
# ============================================================

numeric_columns_cleaned = cleaned_df.select_dtypes(
    include=np.number
).columns

remaining_infinite = np.isinf(
    cleaned_df[numeric_columns_cleaned]
).sum()

print("Remaining infinite values:")
display(remaining_infinite)

if remaining_infinite.sum() == 0:
    print("No infinite values remain.")

# ============================================================
# 24. SAVE CLEANED DATASET
# ============================================================

cleaned_df.to_csv(
    "titanic_cleaned.csv",
    index=False
)

print("Cleaned dataset saved as: titanic_cleaned.csv")

# ============================================================
# 25. EDA - SURVIVAL DISTRIBUTION
# ============================================================

survival_counts = cleaned_df["Survived"].value_counts()

print("Survival counts:")
display(survival_counts)

print("\n0 = Did not survive")
print("1 = Survived")

survival_percentage = (
    cleaned_df["Survived"]
    .value_counts(normalize=True)
    * 100
)

print("Survival percentage:")
display(survival_percentage)

# ============================================================
# 26. VISUALIZATION - SURVIVAL DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=cleaned_df,
    x="Survived"
)

plt.title("Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig(
    "survival_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 27. EDA - SURVIVAL BY GENDER
# ============================================================

gender_survival = pd.crosstab(
    cleaned_df["Sex"],
    cleaned_df["Survived"]
)

print("Survival counts by gender:")
display(gender_survival)

# ============================================================
# 27. EDA - SURVIVAL BY GENDER
# ============================================================

gender_survival = pd.crosstab(
    cleaned_df["Sex"],
    cleaned_df["Survived"]
)

print("Survival counts by gender:")
display(gender_survival)

# ============================================================
# 28. VISUALIZATION - SURVIVAL BY GENDER
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=cleaned_df,
    x="Sex",
    hue="Survived"
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig(
    "survival_by_gender.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 29. EDA - SURVIVAL BY PASSENGER CLASS
# ============================================================

class_survival = pd.crosstab(
    cleaned_df["Pclass"],
    cleaned_df["Survived"]
)

print("Survival counts by passenger class:")
display(class_survival)

class_survival_rate = pd.crosstab(
    cleaned_df["Pclass"],
    cleaned_df["Survived"],
    normalize="index"
) * 100

print("Survival percentage by passenger class:")
display(class_survival_rate)

# ============================================================
# 30. VISUALIZATION - SURVIVAL BY PASSENGER CLASS
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=cleaned_df,
    x="Pclass",
    hue="Survived"
)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig(
    "survival_by_class.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 31. EDA - AGE DISTRIBUTION
# ============================================================

print("Age statistics:")
display(cleaned_df["Age"].describe())

# ============================================================
# 32. VISUALIZATION - AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=cleaned_df,
    x="Age",
    bins=30,
    kde=True
)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig(
    "age_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 33. EDA - FARE DISTRIBUTION
# ============================================================

print("Fare statistics:")
display(cleaned_df["Fare"].describe())

# ============================================================
# 34. VISUALIZATION - FARE DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=cleaned_df,
    x="Fare",
    bins=30,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig(
    "fare_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 35. CREATE FAMILY SIZE FEATURE
# ============================================================

cleaned_df["FamilySize"] = (
    cleaned_df["SibSp"] +
    cleaned_df["Parch"] +
    1
)

display(
    cleaned_df[
        ["SibSp", "Parch", "FamilySize"]
    ].head()
)

# ============================================================
# 36. FAMILY SIZE ANALYSIS
# ============================================================

print("Family size statistics:")
display(cleaned_df["FamilySize"].describe())

# ============================================================
# 37. VISUALIZATION - FAMILY SIZE VS SURVIVAL
# ============================================================

plt.figure(figsize=(10, 5))

sns.countplot(
    data=cleaned_df,
    x="FamilySize",
    hue="Survived"
)

plt.title("Survival by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig(
    "family_size_survival.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 38. CORRELATION ANALYSIS
# ============================================================

numeric_df = cleaned_df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_df.corr()

print("Correlation matrix:")
display(correlation_matrix)

# ============================================================
# 39. VISUALIZATION - CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    "correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 40. AGE GROUP ANALYSIS
# ============================================================

cleaned_df["AgeGroup"] = pd.cut(
    cleaned_df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=[
        "Child",
        "Teenager",
        "Young Adult",
        "Adult",
        "Senior"
    ]
)

age_group_survival = (
    cleaned_df
    .groupby("AgeGroup", observed=True)["Survived"]
    .mean() * 100
)

print("Survival percentage by age group:")
display(age_group_survival)

# ============================================================
# 41. VISUALIZATION - SURVIVAL BY AGE GROUP
# ============================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    x=age_group_survival.index,
    y=age_group_survival.values
)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    "survival_by_age_group.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 42. SURVIVAL BY CLASS AND GENDER
# ============================================================

class_gender_survival = pd.pivot_table(
    cleaned_df,
    values="Survived",
    index="Pclass",
    columns="Sex",
    aggfunc="mean"
) * 100

print("Survival rate by passenger class and gender:")
display(class_gender_survival)

# ============================================================
# 43. OUTLIER CHECK
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=cleaned_df,
    x="Fare"
)

plt.title("Fare Distribution and Outliers")
plt.xlabel("Fare")

plt.tight_layout()

plt.savefig(
    "fare_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 44. FINAL CLEANED DATASET
# ============================================================

print("Final cleaned dataset:")
display(cleaned_df.head())

print("\nFinal shape:")
print(cleaned_df.shape)

print("\nFinal columns:")
print(list(cleaned_df.columns))

# ============================================================
# 45. FINAL DATA QUALITY REPORT
# ============================================================

final_report = pd.DataFrame({
    "Column": cleaned_df.columns,
    "Data Type": cleaned_df.dtypes.astype(str).values,
    "Missing Values": cleaned_df.isnull().sum().values,
    "Unique Values": cleaned_df.nunique().values
})

print("Final Data Quality Report:")
display(final_report)

# ============================================================
# 46. KEY PROJECT STATISTICS
# ============================================================

total_passengers = len(cleaned_df)

survived_count = cleaned_df["Survived"].sum()

survival_rate = (
    cleaned_df["Survived"].mean() * 100
)

female_survival_rate = (
    cleaned_df.loc[
        cleaned_df["Sex"] == "female",
        "Survived"
    ].mean() * 100
)

male_survival_rate = (
    cleaned_df.loc[
        cleaned_df["Sex"] == "male",
        "Survived"
    ].mean() * 100
)

class_survival = (
    cleaned_df.groupby("Pclass")["Survived"]
    .mean() * 100
)

highest_class = class_survival.idxmax()
highest_class_rate = class_survival.max()

print("================================================")
print("KEY PROJECT STATISTICS")
print("================================================")

print(f"Total passengers analyzed: {total_passengers}")
print(f"Passengers who survived: {survived_count}")
print(f"Overall survival rate: {survival_rate:.2f}%")

print(f"\nFemale survival rate: {female_survival_rate:.2f}%")
print(f"Male survival rate: {male_survival_rate:.2f}%")

print(
    f"\nPassenger class with highest survival rate: "
    f"{highest_class}"
)

print(
    f"Highest class survival rate: "
    f"{highest_class_rate:.2f}%"
)

# ============================================================
# 47. KEY INSIGHTS
# ============================================================

print("================================================")
print("KEY INSIGHTS FROM THE ANALYSIS")
print("================================================")

print(
    f"1. The dataset contains {total_passengers} passenger records "
    f"after cleaning."
)

print(
    f"2. The overall survival rate was {survival_rate:.2f}%."
)

print(
    f"3. Female passengers had a survival rate of "
    f"{female_survival_rate:.2f}%, compared with "
    f"{male_survival_rate:.2f}% for male passengers."
)

print(
    f"4. Passenger class {highest_class} had the highest "
    f"survival rate at {highest_class_rate:.2f}%."
)

print(
    "5. Missing values were handled using appropriate "
    "techniques based on variable type."
)

print(
    "6. The Cabin column was removed because it contained "
    "a very high proportion of missing observations."
)

print(
    "7. Numerical variables were examined using descriptive "
    "statistics and correlation analysis."
)

print(
    "8. Multiple visualizations were created to identify "
    "patterns and relationships in the dataset."
)

# ============================================================
# 48. SAVE FINAL CLEANED DATASET
# ============================================================

cleaned_df.to_csv(
    "titanic_cleaned_final.csv",
    index=False
)

print("Final cleaned dataset saved as:")
print("titanic_cleaned_final.csv")

# ============================================================
# 49. VISUALIZATION FILES CREATED
# ============================================================

visualization_files = [
    "missing_values.png",
    "survival_distribution.png",
    "survival_by_gender.png",
    "survival_by_class.png",
    "age_distribution.png",
    "fare_distribution.png",
    "family_size_survival.png",
    "correlation_heatmap.png",
    "survival_by_age_group.png",
    "fare_boxplot.png"
]

print("Visualization files created:")

for file in visualization_files:
    print("✓", file)
# ============================================================
# 50. PROJECT COMPLETION
# ============================================================

print("\n" + "=" * 60)
print("WEEK 1 DATA SCIENCE PROJECT COMPLETED")
print("=" * 60)

print("\nTasks completed:")
print("✓ Data acquisition")
print("✓ Dataset loading")
print("✓ Dataset inspection")
print("✓ Data type analysis")
print("✓ Missing value analysis")
print("✓ Duplicate record analysis")
print("✓ Infinite value analysis")
print("✓ Data cleaning")
print("✓ Statistical analysis")
print("✓ Exploratory Data Analysis")
print("✓ Visualization")
print("✓ Correlation analysis")
print("✓ Key insights")
print("✓ Cleaned dataset export")

print("\nFiles generated:")
print("✓ titanic_cleaned_final.csv")
print("✓ Visualization PNG files")

print("\nYour Week 1 analysis is ready for documentation.")

```
